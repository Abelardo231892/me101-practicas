```python
import pandas as pd
import numpy as np
from scipy import stats

df = pd.read_csv("estudiantes_limpio.csv")

media = df["nota"].mean()
mediana = df["nota"].median()
std_muestral = df["nota"].std(ddof=1)
var_poblacional = np.var(df["nota"], ddof=0)
q1 = df["nota"].quantile(0.25)
q3 = df["nota"].quantile(0.75)
iqr = q3 - q1

print("Media :", media)
print("Mediana :", media)
print("Desviacion estandar MUESTRAL:", std_muestral)
print("Varianza POBLACIONAL :", var_poblacional)
print("Q1 :", q1)
print("Q3 :", q3)
print("IQR :", iqr)
```

```python
skew_pandas = df["nota"].skew()
kurt_pandas = df["nota"].kurt()
skew_scipy = stats.skew(df["nota"].dropna())
kurt_scipy = stats.kurtosis(df["nota"].dropna())

print(f"Pandas -> Skewness: {skew_pandas:.4f}, Kurtosis: {kurt_pandas:.4f}")
print(f"SciPy  -> Skewness: {skew_scipy:.4f}, Kurtosis: {kurt_scipy:.4f}")
```

```python
def resumen_estadistico(serie, decimales=4):
    serie_clean = serie.dropna()
    n = len(serie_clean)
    media_val = serie_clean.mean()
    mediana_val = serie_clean.median()
    desv_std = serie_clean.std(ddof=1)
    cv_pct = (desv_std / media_val) * 100 if media_val != 0 else np.nan
    
    return {
        "n": n,
        "media": round(media_val, decimales),
        "mediana": round(mediana_val, decimales),
        "desv_std": round(desv_std, decimales),
        "cv_pct": round(cv_pct, decimales)
    }

print("Resumen estadístico de nota:", resumen_estadistico(df["nota"]))
```

```python
def clasificar_dispersion(cv_pct):
    if cv_pct < 15:
        return "Baja"
    elif 15 <= cv_pct < 30:
        return "Moderada"
    else:
        return "Alta"

for columna in ["nota", "asistencia_pct"]:
    resumen = resumen_estadistico(df[columna])
    nivel = clasificar_dispersion(resumen["cv_pct"])
    print(f"Columna '{columna}': CV = {resumen['cv_pct']}% -> Dispersión {nivel}")
```
