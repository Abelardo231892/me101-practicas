```python
import numpy as np
notas = np.array([11, 14, 20, 8, 16, 13])
print("Shape:", notas.shape, "dtype :", notas.dtype, "ndim :", notas.ndim)
```

```python
notas = np.array([11, 14, 20, 8, 16, 13])
promedio = notas.mean()
maximo = notas.max()
cantidad_aprobados = (notas>=10.5).sum()

print("Promedio:", promedio)
print("Maximo:", maximo)
print("Aprobados:", cantidad_aprobados)
```

```python
import pandas as pd
datos = {
    "nombre": ["Juan","Roberto","Ana","Yeni","Emily"],
    "nota": [11,20,14,35,7],
    "asistencia_pct": [50,70,80,20,100],
}
df = pd.DataFrame(datos)

df_filtrado = df[(df["nota"] >= 10.5) & (df["asistencia_pct"] >= 70)]
print(df_filtrado)
```
